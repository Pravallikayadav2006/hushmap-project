#Concept of Encoding
What is Encoding?

Machine Learning models cannot understand:

Male
Female
Chicago
Boston
Yes
No

They understand only numbers.

So we convert text into numbers.

This process is called Encoding.

One-Hot Encoding:

Example:

Before:

Gender
Male
Female
Unknown

After:

Gender_Male	Gender_Female	Gender_Unknown
1	0	0
0	1	0
0	0	1
